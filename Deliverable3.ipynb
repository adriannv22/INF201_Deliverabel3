### Dice

Then we have the dice class. It starts with the argument sides. The number of sides has a default value of 6.  

__init__ initialize the dice and the number of sides. The number of sides has to be at least 2. 

The call method returns a random integer when you roll the dice. 

The repr method returns a string representation for exmaple Dice(sides=6)

The call method allows the dice object to be called like a function to roll it. 

```python
#Dice with default number of sides
dice = Dice()

print(dice) 

print(dice.__repr__)
print(dice.roll())

#We use the call method 
print(dice())

#Dice with a chosen number of sides
d20 = Dice(20)

print(d20)

print(dice.__repr__)
print(d20.roll())

#We use the call method
print(d20())
```

### Create items

The weapon class uses the Dice object to decide the damage. It starts with the arguments name and damage_dice. The name is the name of the weapon for example "The legendary sword". damage_dice is the dice object assosciated with the weapon and calculates the damage from the weapon.  

__init__ initializes the weapon with a name and the corresponding dice. 

The roll_damage method rolls the weapons damage_dice and returns the damage value.  


```python
#Create the damage_dice
d15 = Dice(15)

weapon = Weapon("The legendary sword", d15)

print(Weapon)

#We use the weapon and we get the damage value returned. 
weapon.roll_damage()
```
